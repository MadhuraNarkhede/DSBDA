```python
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PowerTransformer
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
data = pd.read_csv('/content/AirQuality.csv',sep=';')
```

```python
data
```

```python
data=data.drop(columns=['Date','Time','Unnamed: 15','Unnamed: 16'])
```

```python
data
```

```python
cols_to_fix=['CO(GT)','C6H6(GT)','T','RH','AH']
```

```python
for column in cols_to_fix:
  data[column]=data[column].str.replace(',','.')
  data[column]=data[column].astype(float)
```

```python
data=data.astype({'CO(GT)':'float64','C6H6(GT)':'float64','T':'float64','RH':'float64','AH':'float64'})
```

```python
data
```

```python
data.dropna(inplace=True)
```

```python
data
```

```python
data.drop_duplicates(inplace=True,ignore_index=True)
```

```python
data
```

```python
def remove_outliers(data,column,lower=0.01,higher=0.99):
  low=data[column].quantile(lower)
  high=data[column].quantile(higher)
  return data[(data[column]>=low)&(data[column]<=high)]
```

```python
df=data.copy()
```

```python
for col in df.columns:
  df=remove_outliers(df,col)
```

```python
data
```

```python
df=df.reset_index(drop=True)
```

```python
data
```

```python
pt=PowerTransformer(method='yeo-johnson')
```

```python
df=pd.DataFrame(pt.fit_transform(df),columns=df.columns)
```

```python
df
```

```python
X=df.drop('AH',axis=1)
```

```python
y=df['AH']
```

```python
X
```

```python
y
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,train_size=0.8,random_state=6,shuffle=True)
```

```python
X_train
```

```python
y_train
```

```python
X_test
```

```python
y_test
```

```python
lr_model=LinearRegression()
lr_model.fit(X_train,y_train)
lr_pred=lr_model.predict(X_test)
```

```python
dt_model=DecisionTreeRegressor()
dt_model.fit(X_train,y_train)
dt_pred=dt_model.predict(X_test)
```

```python
def Evaluate_model(name,y_test,y_pred):
  print(name)
  r2=r2_score(y_test,y_pred)
  mae=mean_absolute_error(y_test,y_pred)
  mse=mean_squared_error(y_test,y_pred)
  print("r2 : ",r2)
  print("mae :",mae)
  print("mse :",mse)
```

```python
Evaluate_model(LinearRegression(),y_test,lr_pred)
```

```python
Evaluate_model(DecisionTreeRegressor(),y_test,lr_pred)
```
