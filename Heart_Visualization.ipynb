```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/heart_disease_uci(1).csv")
```

```python
df
```

```python
df['binary_target']=df['num'].apply(lambda x: 0 if x==0 else 1)
```

```python
df
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='num', y='chol', data=df)
plt.title('Cholesterol Levels by Heart Disease Presence (0 = No, 1 = Yes)')
plt.xlabel('Heart Disease')
plt.ylabel('Cholesterol')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['age'], bins=10, kde=True, color='red')
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
df_sorted = df.sort_values('age')
plt.figure(figsize=(8, 5))
plt.plot(df_sorted['age'], df_sorted['thalch'], color='green')
plt.title('Age vs. Max Heart Rate')
plt.xlabel('Age')
plt.ylabel('Max Heart Rate (thalach)')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(df_sorted['age'], df_sorted['chol'], label='Cholesterol', color='red')
plt.plot(df_sorted['age'], df_sorted['thalch'], label='Max Heart Rate', color='blue')
plt.title('Age vs. Cholesterol and Max Heart Rate')
plt.xlabel('Age')
plt.ylabel('Value')
plt.legend()
plt.grid(True)
plt.show()
```
